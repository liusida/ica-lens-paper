## Paper context

The paper fits each layer on 1,000,000 sampled activations for 50 iterations. On one NVIDIA GB10, the reported per-layer times are 0.92 minutes for GPT-2 Small, 5.5 minutes for Gemma 2 2B, and 12-13 minutes for Qwen 3.5 9B Base.
