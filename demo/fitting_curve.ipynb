# Model-level fitting curves

Compare FastICA fitting for GPT-2 Small, Gemma 2 2B, and Qwen 3.5 9B. Each panel averages each component-percentile boundary across all fitted layers of that model; it does not pool components across layers. Only saved artifact metadata is loaded.

```python
from icalens import ICALens

lenses = {
    "GPT-2 Small": ICALens.from_pretrained("sida/icalens-gpt2-small-pile10k"),
    "Gemma 2 2B": ICALens.from_pretrained("sida/icalens-gemma-2-2b-pile10k"),
    "Qwen 3.5 9B": ICALens.from_pretrained("sida/icalens-qwen3.5-9b-base-pile10k"),
}
lenses["GPT-2 Small"].plot_fitting_curve(layer=0)
```

```python
lenses["GPT-2 Small"].plot_fitting_curve(layers="all", columns=4)
```
