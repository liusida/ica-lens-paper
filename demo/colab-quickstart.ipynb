# ICA Lens: interactive quickstart

ICA Lens discovers and applies interpretable, non-Gaussian directions in language-model activations. This notebook loads a published GPT-2 Lens and inspects the two sentences used in Figure 24 of the paper.

## 1. Set up the runtime

This notebook requests a GPU runtime. Colab assigns the available hardware; a T4 is sufficient. If Colab connects to a CPU runtime, the demo still works but loading and analysis will be slower.

```python
%pip install -q icalens==0.4.0
```

## 2. Load a published Lens

The first call downloads GPT-2 Small and the requested ICA Lens layer from Hugging Face. Later analyses reuse the model in memory.

```python
from icalens import ICALens

lens = ICALens.from_pretrained(
    "sida/icalens-gpt2-small-pile10k",
)
```

## 3. Inspect the library sentence

Run the cell and select components in the interactive result. Expand a component profile to inspect top corpus occurrences and vocabulary readouts.

```python
library = lens.analyze(
    "She arrived at the library to study for her exam.",
    layer=6,
)
library
```

## 4. Inspect the financial-context sentence

Figure 24 uses this sentence to show that C67 follows a broader financial context around **bank**, **deposit**, **check**, **withdraw**, and **cash**, rather than detecting only the surface word *bank*.

```python
financial = lens.analyze(
    "Maya stopped at the bank before the trip, waiting in line to deposit a check and withdraw enough cash for the weekend.",
    layer=6,
)
financial
```
