ICA Lens discovers and applies interpretable, non-Gaussian directions in language-model activations. This notebook loads a published GPT-2 Lens and compares how the same word is represented in two contexts.

## 1. Set up the runtime

For the fastest experience, choose **Runtime → Change runtime type → T4 GPU**. The demo also runs on CPU, but loading and analysis will be slower.

## 2. Load a published Lens

The first call downloads GPT-2 Small and the requested ICA Lens layer from Hugging Face. Later analyses reuse the model in memory.

```python
from icalens import ICALens

lens = ICALens.from_pretrained(
    "sida/icalens-gpt2-small-pile10k",
    device="auto",
)
```

## 3. Inspect a financial context

Run the cell and select components in the interactive result. Expand a component profile to inspect top corpus occurrences and vocabulary readouts.

```python
financial = lens.analyze(
    "She deposited the check at the bank before the trip.",
    layer=6,
)
financial
```

## 4. Compare the same word in a river context

ICA Lens represents a token as a mixture of signed component scores. Compare the components around **bank** with the financial sentence above.

```python
river = lens.analyze(
    "At sunset she sat on the grassy bank and watched the river.",
    layer=6,
)
river
```
