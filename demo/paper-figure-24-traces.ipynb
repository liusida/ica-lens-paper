# Paper demo: Figure 24 sentence-level traces

This notebook reproduces the qualitative GPT-2 Small example from Figure 24. Component C67 follows financial context, while C273 responds to a broader arrival-and-purpose construction. The sentences, layer, and components are taken directly from the paper.

This notebook requests a GPU runtime. Colab assigns available hardware; a T4 is sufficient. It also runs on CPU, more slowly.

```python
from icalens import ICALens

lens = ICALens.from_pretrained(
    "sida/icalens-gpt2-small-pile10k",
)
sentence_1 = "She arrived at the library to study for her exam."
sentence_2 = (
    "Maya stopped at the bank before the trip, waiting in line to deposit "
    "a check and withdraw enough cash for the weekend."
)
first = lens.analyze(sentence_1, layer=6, selected_components=67)
second = lens.analyze(sentence_2, layer=6, selected_components=67)
```

Select C67 or C273 in the interactive explorers and expand **Component profile** to compare token scores with corpus occurrences and vocabulary readouts.

```python
first
```

```python
second
```

```python
import matplotlib.pyplot as plt

components = {67: "Financial", 273: "Arriving/leaving for a purpose"}
fig, axes = plt.subplots(2, 2, figsize=(15, 7), constrained_layout=True)
for column, (result, title) in enumerate(((first, "Sentence 1"), (second, "Sentence 2"))):
    for row, (component, label) in enumerate(components.items()):
        values = result.scores[:, component].detach().cpu().abs().numpy()
        axes[row, column].plot(values, marker="o", color="#136f63")
        axes[row, column].set_xticks(range(len(result.token_texts)), result.token_texts, rotation=65, ha="right")
        axes[row, column].set_title(f"{title}: C{component} ({label})")
        axes[row, column].set_ylabel("|signed ICA score|")
plt.show()
```
