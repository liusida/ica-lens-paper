# ICA component steering

Steer Gemma 2 2B using ICA components and offsets already measured from 1,000 parallel sentence pairs.

```python
import torch

from icalens import ICALens

lens = ICALens.from_pretrained("sida/icalens-gemma-2-2b-pile10k")
prompt = "Artificial intelligence is important because"
layer = 20
generation_kwargs = {
    "max_new_tokens": 50,
    "do_sample": True,
    "temperature": 0.5,
}
```

```python
torch.manual_seed(0)
baseline = lens.generate(prompt, **generation_kwargs)
print("Baseline:", prompt + baseline)
```

## French with current-position steering

```python
French_component = 25
French_delta = 23.079790115356445

torch.manual_seed(0)
steered = lens.generate(
    prompt,
    layer=layer,
    steer=(French_component, French_delta),
    steering_scope="current-position",
    **generation_kwargs,
)
print("French:", prompt + steered)
```

## Chinese with all-position steering

```python
Chinese_component = 105
Chinese_delta = 21.422651290893555

torch.manual_seed(0)
steered = lens.generate(
    prompt,
    layer=layer,
    steer=(Chinese_component, Chinese_delta),
    steering_scope="all-positions", # demonstrate a different scope
    **generation_kwargs,
)
print("Chinese:", prompt + steered)
```
